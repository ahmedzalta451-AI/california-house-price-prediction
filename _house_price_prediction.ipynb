```python
import gradio as gr

import joblib

import numpy as np
```

```python
import joblib

# Save the trained RandomForestRegressor model
joblib.dump(rf_model, "house_price_model.pkl")

# Now, load the model from the saved file
model = joblib.load(
    "house_price_model.pkl"
)
```

```python
def predict_house_price(

    longitude,

    latitude,

    housing_median_age,

    total_rooms,

    total_bedrooms,

    population,

    households,

    median_income,

    ocean_proximity

):


    # One Hot Encoding
    ocean_1h = 0

    inland = 0

    island = 0

    near_bay = 0

    near_ocean = 0



    if ocean_proximity == "<1H OCEAN":
        ocean_1h = 1

    elif ocean_proximity == "INLAND":
        inland = 1

    elif ocean_proximity == "ISLAND":
        island = 1

    elif ocean_proximity == "NEAR BAY":
        near_bay = 1

    elif ocean_proximity == "NEAR OCEAN":
        near_ocean = 1




    # Feature Engineering


    rooms_per_household = total_rooms / households

    bedrooms_per_room = total_bedrooms / total_rooms

    population_per_household = population / households



    # Features Array


    features = np.array([[

        longitude,

        latitude,

        housing_median_age,

        total_rooms,

        total_bedrooms,

        population,

        households,

        median_income,

        ocean_1h,

        inland,

        island,

        near_bay,

        near_ocean,

        rooms_per_household,

        bedrooms_per_room,

        population_per_household

    ]])



    # Prediction

    prediction = model.predict(features)



    # Return Result


    return f"Predicted House Price: ${prediction[0]:,.2f}"
```

```python
inputs = [

    gr.Slider(
        -125,
        -114,
        label="Longitude"
    ),

    gr.Slider(
        32,
        42,
        label="Latitude"
    ),

    gr.Slider(
        1,
        60,
        label="Housing Median Age"
    ),

    gr.Number(
        label="Total Rooms"
    ),

    gr.Number(
        label="Total Bedrooms"
    ),

    gr.Number(
        label="Population"
    ),

    gr.Number(
        label="Households"
    ),

    gr.Number(
        label="Median Income"
    ),

    gr.Dropdown(

        [

            "<1H OCEAN",

            "INLAND",

            "ISLAND",

            "NEAR BAY",

            "NEAR OCEAN"

        ],

        label="Ocean Proximity"

    )

]
```

```python
output = gr.Textbox(

    label="Predicted House Price"

)
```

```python
interface = gr.Interface(

    fn=predict_house_price,

    inputs=inputs,

    outputs=output,



    title="California House Price Prediction",



    description="""

    Predict California house prices using
    Random Forest Regression Machine Learning Model

    """

)
```

```python
interface.launch()
```
